```python
# =====================================================================
# Name: Arinze Ugbah
# Assignment: DATA 201 Week 2 Assignment - Question 8
# Date: September 21, 2026
# =====================================================================

print("==================================================================")
print("--- QUESTION 8: COLLEGE DATASET ---")
print("==================================================================")

# (a) & (b) Load dataset from the class GitHub and set index column
college_url = "https://raw.githubusercontent.com/Reben80/Data201-Fall2026-CRN21732/refs/heads/main/Dataset/College.csv"
college = pd.read_csv(college_url, index_col=0)
print("\n[a & b] First 5 rows of the dataset:")
display(college.head(5))

# (c) Produce a numerical summary of all variables
print("\n[c] Numerical Summary:")
display(college.describe())

# (d) Produce a scatterplot matrix of Top10perc, Apps, and Enroll
print("\n[d] Generating Scatterplot Matrix...")
pd.plotting.scatter_matrix(college[['Top10perc', 'Apps', 'Enroll']], figsize=(10, 10))
plt.show()

# (e) Side-by-side boxplots of Outstate Tuition vs Private/Public indicator
print("\n[e] Generating Outstate vs Private Boxplot...")
college.boxplot(column='Outstate', by='Private', figsize=(8, 6))
plt.title('Out-of-State Tuition by Institutional Type')
plt.suptitle('')
plt.ylabel('Out-of-State Tuition ($)')
plt.show()

# (f) Create and Analyze 'Elite' Universities
print("\n[f] Analyzing Elite Universities...")
college['Elite'] = pd.cut(college['Top10perc'], [0, 50, 100], labels=['No', 'Yes'])
print("Value counts for Elite category:")
print(college['Elite'].value_counts())

college.boxplot(column='Outstate', by='Elite', figsize=(8, 6))
plt.title('Out-of-State Tuition for Elite vs. Non-Elite Universities')
plt.suptitle('')
plt.ylabel('Out-of-State Tuition ($)')
plt.show()

# (g) Histograms with Differing Bins
print("\n[g] Generating 2x2 Histogram Grid...")
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

college['Apps'].plot.hist(ax=axes[0, 0], bins=20, color='skyblue', edgecolor='black')
axes[0, 0].set_title('Histogram of Applications (Bins=20)')

college['Enroll'].plot.hist(ax=axes[0, 1], bins=15, color='salmon', edgecolor='black')
axes[0, 1].set_title('Histogram of Enrollment (Bins=15)')

college['Outstate'].plot.hist(ax=axes[1, 0], bins=25, color='lightgreen', edgecolor='black')
axes[1, 0].set_title('Histogram of Out-of-State Tuition (Bins=25)')

college['Room.Board'].plot.hist(ax=axes[1, 1], bins=10, color='gold', edgecolor='black')
axes[1, 1].set_title('Histogram of Room & Board (Bins=10)')

plt.tight_layout()
plt.show()

# (h)
print("\n[h] MINI ESSAY: SUMMARY OF DISCOVERIES")
print("""
Looking closely at the data and plots, we can see a few clear patterns about
how colleges price themselves and how students apply to them.

* Private Schools Cost Way More: When looking at the tuition chart from part (e),
  it is obvious that private colleges charge much more for out-of-state tuition
  than public universities do. The middle pricing for private schools sits at
  around $12,000, while public state schools usually sit around $7,000.

* Top-Tier Schools Get a Huge Price Premium: In part (f), when we separate colleges
  based on student smarts, 'Elite' schools (where over half of the freshmen came from
  the top 10% of their high school class) charge a lot more money. In fact, even the
  cheapest elite colleges cost almost the same as an average regular college, showing
  that higher academic prestige lets schools charge top dollar.

* A Few Giant Schools Take Most of the Applications: Finally, the histograms show
  that student demand is not split evenly. The vast majority of colleges get a
  small, manageable number of applications (under 5,000). However, a tiny handful
  of massive, well-known universities get an absolute flood of attention, racking up
  well over 20,000 applications each.
""")



```

```python
# =====================================================================
# Student Name: Arinze Ugbah
# Assignment: DATA 201 Week 2 Assignment - Question 9
# Date: September 21, 2026
# =====================================================================

print("==================================================================")
print("--- QUESTION 9: AUTO DATASET ---")
print("==================================================================")

# Setup: Load Auto dataset from GitHub and handle missing values marked with '?'
auto_url = "https://raw.githubusercontent.com/Reben80/Data201-Fall2026-CRN21732/refs/heads/main/Dataset/Auto.csv"
auto = pd.read_csv(auto_url, na_values=['?'])
auto = auto.dropna() # Remove missing observations rows

# (a) PREDICTOR CLASSIFICATION
print("\n[a] PREDICTOR CLASSIFICATION")
print("""
The features in this dataset can be split into two basic groups based on how we count them.
Quantitative features are actual measurements with numbers you can scale up or down, like a
car's gas mileage, engine size, horsepower, weight, and speed. Qualitative features are
group labels or categories, like the car's name, where it was made, its model year, and
how many cylinders it has. Even though features like cylinders or years use numbers, they act
as separate bins to sort the cars rather than a continuous sliding scale.
""")

# Define quantitative variables list for statistical summaries
quant_vars = ['mpg', 'displacement', 'horsepower', 'weight', 'acceleration', 'year']

# (b) & (c) Generate ranges, means, and standard deviations concisely
print("\n[b & c] Summary Statistics for Full Dataset:")
summary_stats = auto[quant_vars].agg(['min', 'max', 'mean', 'std']).T
summary_stats['range'] = summary_stats['max'] - summary_stats['min']
display(summary_stats[['min', 'max', 'range', 'mean', 'std']])

# (d) Subset Analysis (Removing 10th through 85th observations)
print("\n[d] Summary Statistics after dropping 10th through 85th observations:")
auto_subset = auto.drop(auto.index[9:85]) # Slice removing rows 10 to 85 (0-indexed 9 to 85)
subset_stats = auto_subset[quant_vars].agg(['min', 'max', 'mean', 'std']).T
subset_stats['range'] = subset_stats['max'] - subset_stats['min']
display(subset_stats[['min', 'max', 'range', 'mean', 'std']])

# (e) Graphical Analysis (Scatter plot matrix for key performance variables)
print("\n[e] Generating Pairwise Scatterplot Matrix...")
pd.plotting.scatter_matrix(auto[['mpg', 'displacement', 'horsepower', 'weight']], figsize=(10, 10))
plt.show()

# (e) COMMENT ON GRAPHICAL FINDINGS
print("\n[e] COMMENT ON GRAPHICAL FINDINGS")
print("""
When looking at the grid of scatter plots, a few clear trends stand out about
how these car features interact with each other.

* Big, Heavy Cars Get Terrible Gas Mileage: There is a sharp downward curve
  showing that as a car's weight, horsepower, or engine size goes up, its gas
  mileage drops off very fast. The relationship isn't a straight line; the biggest
  drops in efficiency happen when moving from light cars to medium cars, and then
  the trend flattens out for the heaviest vehicles.

* Heavy Cars Naturally Need Big, Powerful Engines: The plots show a straight,
  tight upward line comparing engine size, horsepower, and weight. This makes perfect
  sense in plain terms: heavier cars require larger engine displacements and more
  horsepower just to get moving, meaning these three traits almost always go
  hand-in-hand.
""")

# (f) USEFUL PREDICTORS FOR GAS MILEAGE (mpg)
print("\n[f] USEFUL PREDICTORS FOR GAS MILEAGE (mpg)")
print("""
Yes, the graphs clearly show that weight, horsepower, and engine size are
excellent tools for predicting a car's gas mileage. Because the dots on the charts
cluster tightly along a clear path, knowing any of these details gives you a
very accurate guess of what the fuel efficiency will be.

Additionally, the car's model year is a very useful tracker. If you look at the
plots over time, cars made in later years show a steady upward drift in gas mileage.
This tells us that newer models generally perform better on a gallon of gas, likely
because car technology got better over the years.
""")

```

```python
# =====================================================================
# Student Name: Arinze Ugbah
# Assignment: DATA 201 Week 2 Assignment - Question 10
# Date: September 21, 2026
# =====================================================================

print("==================================================================")
print("--- QUESTION 10: BOSTON HOUSING DATASET ---")
print("==================================================================")

# (a) Load the Boston dataset directly from the provided class GitHub URL
boston_url = "https://raw.githubusercontent.com/Reben80/Data201-Fall2026-CRN21732/refs/heads/main/Dataset/Boston.csv"
boston = pd.read_csv(boston_url)

# If an unnecessary auto-increment index column like 'Unnamed: 0' exists, drop it
if 'Unnamed: 0' in boston.columns:
    boston = boston.drop(columns=['Unnamed: 0'])

# (b) Show exact row and column counts using Python code execution
rows_count = boston.shape[0]
cols_count = boston.shape[1]

print("\n[b] CODE OUTPUT:")
print(f"Number of rows in the dataset: {rows_count}")
print(f"Number of columns in the dataset: {cols_count}")

# (c) Pairwise Relationships (Housing age vs distance to employment centers)
print("\n[c] Generating Housing Age vs. Distance Scatterplot...")
plt.scatter(boston['age'], boston['dis'], alpha=0.5)
plt.xlabel('Proportion of units built prior to 1940 (age)')
plt.ylabel('Weighted distances to 5 employment centres (dis)')
plt.title('Housing Age vs Distance to Employment Centers')
plt.show()

# (c) Pairwise Findings Explanation in Layman's Terms
print("\n[c] COMMENT ON PAIRWISE RELATIONSHIPS:")
print("""
Looking at the chart comparing home age to business hub distance, we can see a clear trend.
Neighborhoods made up almost entirely of old, historic homes are packed close to the city center.
As you move further out into the suburbs, the houses get noticeably newer and modern.
This means outer communities completely lack the historic architecture found near the urban core.
""")

# (d) Check linear correlations with the per capita crime rate ('crim') using code
print("\n[d] CODE OUTPUT: Top factors correlated with crime rate ('crim'):")
crime_correlations = boston.corr()['crim'].sort_values(ascending=False)
display(crime_correlations)

# (d) Crime Association Explanation in Layman's Terms
print("\n[d] PER CAPITA CRIME RATE ASSOCIATIONS:")
print("""
The data shows that crime rates are much higher in areas with lots of industrial businesses, old houses, and high property taxes.
There is also a very strong link between crime and lower home values.
Neighborhoods with the cheapest homes see massive spikes in crime, while the wealthier suburbs stay incredibly safe.
""")

# (e) Suburbs with Unusual Crime, Tax, or Pupil-Teacher Ratios
print("\n[e] Extreme Outlier Suburbs Analysis:")
print("Top 3 Highest Crime Rates:\n", boston['crim'].sort_values(ascending=False).head(3))
print("\nTop 3 Highest Tax Rates:\n", boston['tax'].sort_values(ascending=False).head(3))
print("\nTop 3 Highest Pupil-Teacher Ratios:\n", boston['ptratio'].sort_values(ascending=False).head(3))

# (e) Outlier Commentary in Layman's Terms
print("\n[e] COMMENT ON RANGES AND OUTLIERS:")
print("""
A few specific neighborhoods stand out as extreme exceptions in this dataset.
While most areas have very low crime, the worst neighborhood hits a shocking peak over 88%.
Property taxes also show a strange leap where a large group of towns all hit the exact same high tax cap of 711.
School crowding follows a similar pattern, with multiple communities maxing out at a 22-to-1 student-to-teacher ratio.
""")

# (f) Charles River Proximity (Indicator 'chas' equals 1)
river_count = (boston['chas'] == 1).sum()
print(f"\n[f] Suburbs bounding the Charles River: {river_count}")

# (g) Median Pupil-Teacher Ratio
print(f"\n[g] Median Pupil-Teacher Ratio: {boston['ptratio'].median()}")

# (h) Suburb with Lowest Median Home Value (medv)
lowest_medv_idx = boston['medv'].idxmin()
print(f"\n[h] Suburb index with lowest median home value: {lowest_medv_idx}")
display(boston.iloc[[lowest_medv_idx]])

# (h) Lowest Value Suburb Commentary in Layman's Terms
print("\n[h] LOW VALUE SUBURB ANALYSIS:")
print("""
The neighborhoods with the cheapest houses in the dataset are in much worse shape than the rest of the city.
Every single home in these specific areas was built before 1940, and crime rates hover around a high 40%.
On top of that, residents face heavy property taxes and crowded schools with over 20 students per teacher.
""")

# (i) High-Room Dwellings (> 7 and > 8 rooms)
seven_rooms = (boston['rm'] > 7).sum()
eight_rooms = (boston['rm'] > 8).sum()
print(f"\n[i] Suburbs averaging more than 7 rooms: {seven_rooms}")
print(f"Suburbs averaging more than 8 rooms: {eight_rooms}")

print("\nSummary statistics of suburbs with more than 8 rooms:")
display(boston[boston['rm'] > 8].describe())

# (i) Large Room Commentary in Layman's Terms
print("\n[i] COMMENT ON LARGE ROOM SUBURBS:")
print("""
The neighborhoods that average more than eight rooms per house represent the most premium parts of Boston.
The average home value here is nearly triple the city baseline, and crime rates sit safely under 1%.
These wealthy communities also feature excellent schools with an uncrowded 14-to-1 student-to-teacher ratio.
""")
```
